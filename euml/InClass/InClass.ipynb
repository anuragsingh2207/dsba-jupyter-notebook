```python
def plot_all_boxplots(df_num):
    """Generates box plots for all numeric columns in a DataFrame."""
    
    for col in df_num.columns:
        plt.figure(figsize=(8, 6))  # Adjust figure size if needed
        plt.boxplot(df_num[col].dropna(), patch_artist=True, orient='h')  # Using patch_artist for filled boxes
        plt.title(f'Boxplot for {col}')
        plt.ylabel(col) # Add y-axis label
        plt.grid(axis='y', linestyle='--') # Add a grid for better readability
        plt.show()
        
plot_all_boxplots(df_num)
```

```python
def outlier_treatment(col):
    Q1, Q3 = np.percentile(col, [25, 75])
    IQR = Q3 - Q1
    lower_range = Q1 - (1.5 * IQR)
    upper_range = Q3 + (1.5 * IQR)
    return lower_range, upper_range
```

```python
df_num['Fare'] = np.where(df_num['Fare'] > 350, 350, df[col])
```

```python
for col in df_num.columns:
    lower_range, upper_range = outlier_treatment(df_num[col])
    print(col,lower_range, upper_range )
    
    df_num[col] = np.where(df_num[col] > upper_range, upper_range, df[col])
    df_num[col] = np.where(df_num[col] < lower_range, lower_range, df[col])
```
